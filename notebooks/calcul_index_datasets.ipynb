```python
def indices_to_intervals(indices):
    """
    Convertit une liste d'indices triés en intervalles
    Ex: [1,2,3,8,9,10,15] -> [(1,3), (8,10), (15,15)]
    """
    if not indices:
        return []
    
    intervals = []
    start = indices[0]
    end = indices[0]
    
    for i in range(1, len(indices)):
        if indices[i] == end + 1:
            # Continue l'intervalle actuel
            end = indices[i]
        else:
            # Termine l'intervalle actuel et en commence un nouveau
            intervals.append((start, end))
            start = indices[i]
            end = indices[i]
    
    # Ajouter le dernier intervalle
    intervals.append((start, end))
    
    return intervals


def format_intervals(intervals):
    """
    Formate les intervalles pour affichage
    Ex: [(1,3), (8,10), (15,15)] -> "[1-3] U [8-10] U [15]"
    """
    formatted = []
    for start, end in intervals:
        if start == end:
            formatted.append(f"[{start}]")
        else:
            formatted.append(f"[{start}-{end}]")
    
    return " U ".join(formatted)

def format_intervals_bash(intervals):
    """
    Formate les intervalles pour affichage
    Ex: [(1,3), (8,10), (15,15)] -> "[1-3] U [8-10] U [15]"
    """
    formatted = ""
    for start, end in intervals:
        if start == end:
            formatted += f" \"{start} {end}\" "
        else:
            formatted += f" \"{start} {end}\""
    
    return formatted



intervals = indices_to_intervals(INDEXES_SORT)
#print(intervals)  # [(1, 3), (8, 10), (15, 15), (20, 23)]
print(format_intervals_bash(intervals))  # [1-3] U [8-10] U [15] U [20-23]
```
