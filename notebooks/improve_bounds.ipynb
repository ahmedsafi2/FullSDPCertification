```python
tabs = [tab_1, tab_2, tab_3, tab_4, tab_5, tab_6, tab_7, tab_8, tab_9, tab_10]
ub_cols = [f"UB_Layer_0_Neuron_{j}" for j in range(784)]

indexed = [tab.set_index('data_index')[ub_cols] for tab in tabs]
stacked = pd.concat(indexed).groupby(level='data_index')

ub_max = stacked.max()   # shape: (n_data_index, 784)
ub_min = stacked.min()

max_diff = ub_max - ub_min  # différence max entre deux tabs, par (data_index, neuron)


# Différence max globale (sur tous les neurones et data_index)
print(max_diff.max().max())

# Top neurones les plus variables (en moyenne sur les samples)
print(max_diff.mean().nlargest(10))

# Top (data_index, neuron) : les paires les plus variables
print(max_diff.stack().nlargest(10))
```
